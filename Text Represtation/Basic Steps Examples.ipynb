```python
import pandas as pd
```

```python
df = pd.DataFrame([
    ['people watch campusX' , 1], 
    ['campusX watch campusX' , 1], 
    ['people write comment' , 0], 
    ['campusX write comment' , 0]
] , columns = ['text' , 'output'])
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer 
cv = CountVectorizer()
```

BAG OF WORDS

```python
bow = cv.fit_transform(df['text'])
```

```python
cv.vocabulary_
```

```python
bow[0]
```

```python
bow.toarray()
```

```python
cv.transform(["campusX people write comment campusX"]).toarray()
```

```python
cv.transform(["not in vocab"]).toarray()
```

New words not give error just not included

N GRAMS 

Bag of words is a subclass of N-grams in the case of (1,1)

```python
bi_gram =CountVectorizer(ngram_range=(2,2))
bow = bi_gram.fit_transform(df['text'])
bow.toarray()
```

```python
bi_gram.vocabulary_
```

```python
bi_gram = CountVectorizer(ngram_range=(1,3))
bow = bi_gram.fit_transform(df["text"])
bow.toarray()
```

```python
bi_gram.vocabulary_
```
