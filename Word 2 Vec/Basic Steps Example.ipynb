```python
import gensim
import os
```

```python
import os

story = []
for file in os.listdir('data'):
    with open(os.path.join('data', file), encoding='latin-1') as f:
        story.append(f.read())

len(story)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
without_stopwords = []
stopword = set(stopwords.words('english'))

for text in story:
    sentences = sent_tokenize(text)
    
    for sentence in sentences:
        words = word_tokenize(sentence.lower())
        filtered_sentence = []
        
        for word in words:
            if word not in stopword and word.isalpha():
                filtered_sentence.append(word)
                
        if filtered_sentence:  # ✅ only append if not empty
            without_stopwords.append(filtered_sentence)
        
            
        
```

```python
without_stopwords
```

```python
model = gensim.models.Word2Vec(window =10, min_count =2)

#making vocab with the unique words
model.build_vocab(without_stopwords)

#traing the model
model.train(without_stopwords, total_examples =model.corpus_count, epochs =5)
```

```python
model.wv.most_similar('king')
```
