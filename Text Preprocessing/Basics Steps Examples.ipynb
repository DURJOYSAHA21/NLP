URL REMOVE


```python
def remove_url(text):
    pattern=re.compile(r'https?://\S+|www\.\S+')
    return re.sub(pattern,'',text)
```

```python
text1 = 'Check out my notebook https://www.kaggle.com/campusx/notebook8223fc1abb'
text2 = 'Check out my notebook http://www.kaggle.com/campusx/notebook8223fc1abb'
text3 = 'Google search here www.google.com'
text4 = 'For notebook click https://www.kaggle.com/campusx/notebook8223fc1abb to search check www.google.com'
```

```python
print(remove_url(text1))
print(remove_url(text2))
print(remove_url(text3))
print(remove_url(text4))
```

REMOVE PUNCTUATIONS

```python
import string
string.punctuation
```

```python
exclude=string.punctuation
```

```python
def remove_punc(text):
    return text.translate(str.maketrans('','',exclude))
```

Note: maketrans(Characters to replace, Characters to replace them with, Characters to delete)
'': no character to replace
'': no character to replace with
punc_list: characters to delete from the text

STOP WORDS REMOVE


```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stopwords.words('bengali')
```

```python
def stop_words(text):
    new_text=[]
    stop_words=set(stopwords.words('english'))
    for word in text.split():
        if word in stop_words:
            new_text.append('')
        else:
            new_text.append(word)
        
    return " ".join(new_text)
            
```

```python
stop_words("Basics Steps Examples.ipynb is a notebook which contains the basic steps of text preprocessing in NLP")
```

```python
df['review']=df['review'].apply(stop_words)
df.head()
```

```python
def remove_extra_space(text):
    return " ".join(text.split())
df['review']=df['review'].apply(remove_extra_space)
df.head()
```

REMOVE EMOJI

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

```python
remove_emoji("Lmao 😂😂")
```

STEMMING

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stemming(text):
    stemmed_text=[]
    for word in text.split():
        stemmed_text.append(ps.stem(word))
    return " ".join(stemmed_text)
```

```python
text="Basics Steps Examples.ipynb is a notebook which contains the basic steps of text preprocessingin NLP. It is a very useful notebook for beginners! Do you like it?"
stemming(text)
```

```python
sample = "walk walks walking walked"
stemming(sample)
```

LEMMATIZER

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemma = WordNetLemmatizer()
punctuation = string.punctuation

def lemmatization(text):
    token=nltk.word_tokenize(text)
    for word in token:
        if word in punctuation:
            token.remove(word)
    print("{0:20}{1:20}".format("Word","Lemma"))
    for word in token:
        print("{0:20}{1:20}".format(word,lemma.lemmatize(word,pos='v')))
```

```python
text="Basics running Steps Examples.ipynb is a notebook which contains the basic steps of text preprocessingin NLP. It is a very useful notebook for beginners! Do you like it?"
lemmatization(text)
```
